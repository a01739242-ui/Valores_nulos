```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
file_path = "/Users/ximenamarin/Desktop/valores_nulos/Valores_nulos/actividad2.1/Citas_Digital_Filtrado.csv"
data = pd.read_csv(file_path)
data.head(5)
```

```python
#Eliminación de la columna 'Unnamed: 15' del DataFrame
data1= data.drop(['Unnamed: 15'], axis=1)
data1
```

```python
#Encontrar el total de valores nulos en cada columna del DataFrame
valores_nulos=data1.isnull().sum()
valores_nulos
```

```python
#Identificar el total de valores nulos en todo el DataFrame
valores_nulos=data1.isnull().sum().sum()
valores_nulos
```

```python
data1.dtypes
```

```python
#Inicio con el tratamiento de los Valores Nulos
data2 = data1.copy()
```

```python
data2.columns
```

```python
#Valores nulos de la columna BDC con por valores no nulos hacia adelante  "forward fill" ("ffill")
data2['BDC '] = data2['BDC '].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2.shape
```

```python
#Sustituimos los valores nulos de la columna Fecha por 0
data2["Fecha"]= data2["Fecha"].fillna(0) 
data2.head(2)
```

```python
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2['Nombre Cliente '] = data2['Nombre Cliente '].fillna("Sin dato")
```

```python
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2["Telefono "]= data2["Telefono "].fillna(0) 
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2['Estatus de Lead'] = data2['Estatus de Lead'].fillna("Sin dato")
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
data2['Potencial de compra'] = data2['Potencial de compra'].astype(float)
```

```python
data2['Potencial de compra']=data2['Potencial de compra'].astype("string")
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("Ventas ","0", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("%","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("'","", regex=False)
data2['Potencial de compra']=data2['Potencial de compra'].str.replace("&","", regex=False)
```

```python
data2["Potencial de compra"] = data2["Potencial de compra"].fillna(round(data2["Potencial de compra"].median(),1))
```
